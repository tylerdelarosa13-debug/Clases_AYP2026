```python
indice_1 = ("lunes", "martes", "miércoles")
marco_1 = pd.DataFrame({'Centro':[1520,1640,1043],
                       'Sur':[1422,1673,1534],
                      'Oriente':[1021,1073,1100]},
                       index=indice_1)
indice_1
marco_1
```

```python
indice_2 = ("jueves", "viernes", "sábado")
marco_2 = pd.DataFrame({'Centro':[1321,1459,1875],
                       'Sur':[1622,1841,1920],
                      'Oriente':[1500,1432,1491]},
                      index=indice_2)

indice_2
marco_2
```
