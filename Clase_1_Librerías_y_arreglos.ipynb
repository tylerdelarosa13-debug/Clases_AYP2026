```python
import time
lista = list(range(1000000))
inicio = time.time()
[valor * 2 for valor in lista]
print("Tiempo de ejecucion con listas", time.time()-inicio)
```

```python
import numpy as np
arreglo = np.arange(1000000)
inicio = time.time()
arreglo * 2
print("Tiempo de ejecucion con listas", time.time() - inicio)
```
